```python
import pandas as pd

df = pd.read_excel(
    r"C:\Users\HYPE AMD\Downloads\Dataset nilai siswa.xlsx"
)

df.head()
```

```python
import time

start_time = time.perf_counter()
```

```python
sains = [
    "Matematika",
    "Fisika",
    "Biologi",
    "Kimia"
]

sosial = [
    "Sejarah",
    "PKN",
    "Ekonomi",
    "Sosiologi"
]

bahasa = [
    "Bahasa Inggris",
    "Bahasa Indonesia"
]

semua_mapel = sains + sosial + bahasa
```

```python
df["Rata_Rata"] = df[semua_mapel].mean(axis=1)

df[
    ["Nama", "Rata_Rata"]
].head()
```

```python
df["Rata_Sains"] = df[sains].mean(axis=1)

df["Rata_Sosial"] = df[sosial].mean(axis=1)

df["Rata_Bahasa"] = df[bahasa].mean(axis=1)

df[
    [
        "Nama",
        "Rata_Sains",
        "Rata_Sosial",
        "Rata_Bahasa"
    ]
].head()
```

```python
def bidang_dominan(row):

    nilai = {
        "Sains": row["Rata_Sains"],
        "Sosial": row["Rata_Sosial"],
        "Bahasa": row["Rata_Bahasa"]
    }

    return max(nilai, key=nilai.get)

df["Bidang_Dominan"] = df.apply(
    bidang_dominan,
    axis=1
)

df[
    [
        "Nama",
        "Bidang_Dominan"
    ]
].head()
```

```python
def rekomendasi_jurusan(bidang):

    if bidang == "Sains":
        return "Teknik Informatika"

    elif bidang == "Sosial":
        return "Manajemen"

    else:
        return "Sastra Inggris"

df["Rekomendasi_Jurusan"] = (
    df["Bidang_Dominan"]
    .apply(rekomendasi_jurusan)
)
```

```python
KKM = 75

df["Status"] = df["Rata_Rata"].apply(
    lambda x:
    "Lulus"
    if x >= KKM
    else "Tidak Lulus"
)
```

```python
tertinggi = df.loc[
    df["Rata_Rata"].idxmax()
]

terendah = df.loc[
    df["Rata_Rata"].idxmin()
]

print("Siswa Terbaik")
print(
    tertinggi["Nama"],
    round(tertinggi["Rata_Rata"],2)
)

print()

print("Siswa Terendah")
print(
    terendah["Nama"],
    round(terendah["Rata_Rata"],2)
)
```
